# FitGuard 启动自检
摄像头和ESP32未连接时也可以运行；此时应显示 offline_ready=true、live_ready=false。

```python
from pathlib import Path
import importlib, sys
deploy = Path.cwd().resolve()
base = (deploy.parent / 'FitGuard_Unified_Test').resolve()
for path in (deploy, base):
    if str(path) not in sys.path: sys.path.insert(0, str(path))
import fitguard_preflight
fitguard_preflight = importlib.reload(fitguard_preflight)
report = fitguard_preflight.run_preflight(deploy, base)
report
```
