# FitGuard 无线 IMU 独立验收
本页只读取 `/dev/ttyUSB0`，不加载 FPGA、不启动摄像头、不修改原数据集。

```python
from pathlib import Path
import sys
base = (Path.cwd().parent / 'FitGuard_Unified_Test').resolve()
if str(base) not in sys.path:
    sys.path.insert(0, str(base))
from fitguard_gateway_imu import GatewayMPU6050
print('基础工程:', base)
print('候选串口:', GatewayMPU6050.available_ports())
imu = GatewayMPU6050(port='/dev/ttyUSB0', baud=921600)
try:
    identity = imu.initialize(timeout_s=5.0)
    print(f'无线 IMU 已连接，WHO_AM_I=0x{identity:02X}')
    print('请保持传感器静止，正在接收约 5 秒数据……')
    samples = [imu.read_sample(timeout=0.5) for _ in range(250)]
    print('最后一条样本:', samples[-1])
    print('链路统计:', imu.diagnostics())
finally:
    imu.close()
```
