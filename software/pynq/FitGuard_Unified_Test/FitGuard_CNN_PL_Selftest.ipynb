# FitGuard CNN-PL 独立验收
不需要摄像头或无线 IMU。加载新增 bitstream，通过第二路 DMA 跑 18 组金样，检查数值误差与实际板端延迟。

```python
from pathlib import Path
import subprocess, sys
base = Path.cwd().resolve()
cmd = [sys.executable, str(base / 'fitguard_cnn_pl_selftest.py'),
       '--bitstream', str(base / 'fitguard_unified_cnn_pl.bit'),
       '--golden', str(base / 'fitguard_cnn_pl_golden.npz'),
       '--report', str(base / 'fitguard_cnn_pl_board_selftest.json')]
completed = subprocess.run(cmd, text=True, stdout=subprocess.PIPE, stderr=subprocess.STDOUT)
print(completed.stdout)
if completed.returncode:
    raise RuntimeError(f'CNN-PL self-test failed with code {completed.returncode}')
```
